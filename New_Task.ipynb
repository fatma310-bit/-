```python
records = [
{"date": "2026-09-25", "category": "food",  "amount_egp": 85.5,
"payment_method": "cash",  "quantity": 1},
{"date": "2026-09-25", "category": "drink", "amount_egp": 40.0,
"payment_method": "card",  "quantity": 2},
{"date": "2026-09-26", "category": "food",  "amount_egp": 120.0,
"payment_method": "cash", "quantity": 3},
# TODO: ﻛﻤﻞ ﻟﺤﺪ 10 ﺳﺠﻼت ﺑﻘﯿﻢ واﻗﻌﯿﺔ ﻣﻦ ﻣﻮﺿﻮﻋﻚ
]
```

```python
FIELDS = ("date", "category", "amount_egp", "payment_method", "quantity")
# TODO: ﻋﺪل اﻷﺳﻤﺎء ﻟﺤﻘﻞ ﻛﺎرت ﻣﺸﺮوﻋﻚ أﻧﺖ ﻟﻮ ﻣﺨﺘﻠﻔﯿﻦ
categories = set(r["category"] for r in records)   # ﻛﻮد ﺟﺎھﺰ: ﻗﯿﻢ ا
```
