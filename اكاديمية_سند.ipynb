```python
import pandas as pd

students = pd.DataFrame({
    "م": [1, 2, 3, 4],
    "اسم المعلم/المعلمة": ["أستاذ أحمد", "أستاذ أحمد", "أستاذة سارة", "أستاذة سارة"],
    "اسم الطالب": ["أحمد محمد", "محمد علي", "مريم أحمد", "سارة محمود"],
    "رقم الحلقة": [1, 1, 2, 2],
    "الدوري الأول": [0, 0, 0, 0],
    "نصف العام": [0, 0, 0, 0],
    "الدوري الثاني": [0, 0, 0, 0],
    "آخر العام": [0, 0, 0, 0],
    "التفسير الأول": [0, 0, 0, 0],
    "التفسير الثاني": [0, 0, 0, 0],
    "التجويد": [0, 0, 0, 0],
    "الغياب": [0, 0, 0, 0],
    "السلوك": [0, 0, 0, 0]
})

students
```

```python
students["المجموع"] = (
    students["الدوري الأول"] +
    students["نصف العام"] +
    students["الدوري الثاني"] +
    students["آخر العام"] +
    students["التفسير الأول"] +
    students["التفسير الثاني"] +
    students["التجويد"] +
    students["الغياب"] +
    students["السلوك"]
)

students[["اسم الطالب", "المجموع"]]
```

```python
def get_grade(score):
    if score >= 90:
        return "ممتاز"
    elif score >= 80:
        return "جيد جدًا"
    elif score >= 70:
        return "جيد"
    elif score >= 60:
        return "مقبول"
    else:
        return "يحتاج متابعة"

students["التقدير"] = students["المجموع"].apply(get_grade)

students[["اسم الطالب", "المجموع", "التقدير"]]
```

```python
import ipywidgets as widgets
from IPython.display import display

name = widgets.Text(
    description="اسم الطالب:",
    placeholder="اكتبي اسم الطالب"
)

ring = widgets.IntText(
    description="رقم الحلقة:",
    value=1
)

teacher = widgets.Text(
    description="اسم المعلم:",
    placeholder="اسم المعلم أو المعلمة"
)

add_button = widgets.Button(
    description="إضافة الطالب"
)

output = widgets.Output()

def add_student(b):
    with output:
        print(
            "تمت إضافة:",
            name.value,
            "— الحلقة:",
            ring.value,
            "— المعلم:",
            teacher.value
        )

add_button.on_click(add_student)

display(name, ring, teacher, add_button, output)
```

```python
def add_student_to_table(b):
    global students

    new_student = {
        "م": len(students) + 1,
        "اسم الطالب": name.value,
        "رقم الحلقة": ring.value,
        "اسم المعلم/المعلمة": teacher.value,
        "الدوري الأول": 0,
        "نصف العام": 0,
        "الدوري الثاني": 0,
        "آخر العام": 0,
        "التفسير الأول": 0,
        "التفسير الثاني": 0,
        "التجويد": 0,
        "الغياب": 0,
        "السلوك": 0
    }

    students.loc[len(students)] = new_student

    print("تمت إضافة الطالب بنجاح ✅")
    display(students)

add_button.on_click(add_student_to_table)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# خانات اختيار الطالب
student_selector = widgets.Dropdown(
    options=list(students["اسم الطالب"]),
    description="الطالب:"
)

exam_selector = widgets.Dropdown(
    options=[
        "الدوري الأول",
        "نصف العام",
        "الدوري الثاني",
        "آخر العام",
        "التفسير الأول",
        "التفسير الثاني",
        "التجويد",
        "الغياب",
        "السلوك"
    ],
    description="الدرجة:"
)

score_input = widgets.IntText(
    value=0,
    description="الدرجة:"
)

save_score = widgets.Button(
    description="حفظ الدرجة"
)

result = widgets.Output()

def save_student_score(b):
    with result:
        clear_output()

        student_name = student_selector.value
        exam_name = exam_selector.value
        score = score_input.value

        # البحث عن الطالب
        index = students.index[
            students["اسم الطالب"] == student_name
        ][0]

        # حفظ الدرجة
        students.loc[index, exam_name] = score

        # إعادة حساب المجموع
        students.loc[index, "المجموع"] = (
            students.loc[index, "الدوري الأول"] +
            students.loc[index, "نصف العام"] +
            students.loc[index, "الدوري الثاني"] +
            students.loc[index, "آخر العام"] +
            students.loc[index, "التفسير الأول"] +
            students.loc[index, "التفسير الثاني"] +
            students.loc[index, "التجويد"] +
            students.loc[index, "الغياب"] +
            students.loc[index, "السلوك"]
        )

        # إعادة حساب التقدير
        students.loc[index, "التقدير"] = get_grade(
            students.loc[index, "المجموع"]
        )

        print("تم حفظ الدرجة بنجاح ✅")
        display(
            students[
                students["اسم الطالب"] == student_name
            ][[
                "اسم الطالب",
                "الدوري الأول",
                "نصف العام",
                "الدوري الثاني",
                "آخر العام",
                "التفسير الأول",
                "التفسير الثاني",
                "التجويد",
                "الغياب",
                "السلوك",
                "المجموع",
                "التقدير"
            ]]
        )

save_score.on_click(save_student_score)

display(
    student_selector,
    exam_selector,
    score_input,
    save_score,
    result
)
```

```python
# شاشة إضافة طالب جديدة

def show_add_student(b=None):
    clear_output(wait=True)

    title_add = widgets.HTML(
        value="""
        <h2 style="text-align:center;">
            ➕ إضافة طالب جديد
        </h2>
        """
    )

    student_name = widgets.Text(
        description="اسم الطالب:",
        placeholder="اكتبي اسم الطالب"
    )

    ring_number = widgets.IntText(
        description="رقم الحلقة:",
        value=1
    )

    teacher_name = widgets.Text(
        description="المعلم/المعلمة:",
        placeholder="اكتبي اسم المعلم أو المعلمة"
    )

    add_button = widgets.Button(
        description="💾 حفظ الطالب",
        button_style="success"
    )

    back_button = widgets.Button(
        description="⬅️ الرئيسية"
    )

    message = widgets.Output()

    def add_new_student(b):
        global students

        with message:
            clear_output()

            if student_name.value.strip() == "":
                print("⚠️ من فضلك اكتبي اسم الطالب.")
                return

            new_student = {
                "م": len(students) + 1,
                "اسم الطالب": student_name.value.strip(),
                "رقم الحلقة": ring_number.value,
                "اسم المعلم/المعلمة": teacher_name.value.strip(),
                "الدوري الأول": 0,
                "نصف العام": 0,
                "الدوري الثاني": 0,
                "آخر العام": 0,
                "التفسير الأول": 0,
                "التفسير الثاني": 0,
                "التجويد": 0,
                "الغياب": 0,
                "السلوك": 0,
                "المجموع": 0,
                "التقدير": "يحتاج متابعة"
            }

            students.loc[len(students)] = new_student

            print("✅ تم إضافة الطالب بنجاح!")

            student_name.value = ""
            teacher_name.value = ""

    add_button.on_click(add_new_student)
    back_button.on_click(show_home)

    display(title_add)
    display(student_name)
    display(ring_number)
    display(teacher_name)
    display(add_button)
    display(back_button)
    display(message)
```

```python
def search_students(b=None):
    search_text = search_box.value.strip()

    with search_result:
        clear_output()

        if search_text == "":
            display(students)
            return

        result = students[
            students["اسم الطالب"].astype(str).str.contains(
                search_text,
                case=False,
                na=False
            )
        ]

        if result.empty:
            print("❌ لم يتم العثور على الطالب.")
        else:
            display(result)


search_box = widgets.Text(
    description="🔎 بحث:",
    placeholder="اكتبي اسم الطالب"
)

search_result = widgets.Output()

search_button = widgets.Button(
    description="بحث",
    layout=widgets.Layout(width="100px")
)

search_button.on_click(search_students)
```

```python
def show_students_final(b=None):
    clear_output(wait=True)

    title_students = widgets.HTML(
        value="""
        <h2 style="text-align:center;">
            👨‍🎓 إدارة الطلاب
        </h2>
        """
    )

    search_box = widgets.Text(
        description="🔎 بحث:",
        placeholder="اكتبي اسم الطالب"
    )

    search_btn = widgets.Button(
        description="بحث",
        layout=widgets.Layout(width="100px")
    )

    add_btn = widgets.Button(
        description="➕ إضافة طالب",
        button_style="success",
        layout=widgets.Layout(width="160px")
    )

    all_btn = widgets.Button(
        description="📋 كل الطلاب",
        layout=widgets.Layout(width="140px")
    )

    back_btn = widgets.Button(
        description="⬅️ الرئيسية",
        layout=widgets.Layout(width="140px")
    )

    result = widgets.Output()

    def search(b):
        with result:
            clear_output()

            text = search_box.value.strip()

            if text == "":
                display(students)
                return

            found = students[
                students["اسم الطالب"].astype(str).str.contains(
                    text,
                    case=False,
                    na=False
                )
            ]

            if found.empty:
                print("❌ لم يتم العثور على الطالب.")
            else:
                display(found)

    def show_all(b):
        with result:
            clear_output()
            display(students)

    search_btn.on_click(search)
    all_btn.on_click(show_all)
    add_btn.on_click(show_add_student)
    back_btn.on_click(show_home)

    display(title_students)

    display(
        widgets.HBox([
            search_box,
            search_btn
        ])
    )

    display(
        widgets.HBox([
            add_btn,
            all_btn,
            back_btn
        ])
    )

    display(result)

    # عرض كل الطلاب عند فتح الشاشة
    with result:
        display(students)


# ربط زر الطلاب بالواجهة الجديدة
students_btn._click_handlers.callbacks = []
students_btn.on_click(show_students_final)

# عرض الرئيسية
clear_output(wait=True)
display(title)
display(menu)
```

```python
def show_grades(b=None):
    clear_output(wait=True)

    title_grades = widgets.HTML(
        value="""
        <h2 style="text-align:center;">
            📝 درجات الطلاب
        </h2>
        """
    )

    student_box = widgets.Dropdown(
        options=list(students["اسم الطالب"]),
        description="الطالب:"
    )

    exam_box = widgets.Dropdown(
        options=[
            "الدوري الأول",
            "نصف العام",
            "الدوري الثاني",
            "آخر العام",
            "التفسير الأول",
            "التفسير الثاني",
            "التجويد",
            "الغياب",
            "السلوك"
        ],
        description="المادة:"
    )

    score_box = widgets.IntText(
        value=0,
        description="الدرجة:"
    )

    save_btn = widgets.Button(
        description="💾 حفظ الدرجة",
        button_style="success"
    )

    back_btn = widgets.Button(
        description="⬅️ الرئيسية"
    )

    result = widgets.Output()

    def save_grade(b):
        with result:
            clear_output()

            if len(students) == 0:
                print("❌ لا يوجد طلاب.")
                return

            student_name = student_box.value
            exam_name = exam_box.value
            score = score_box.value

            # الحدود القصوى للدرجات
            max_scores = {
                "الدوري الأول": 40,
                "نصف العام": 100,
                "الدوري الثاني": 40,
                "آخر العام": 100,
                "التفسير الأول": 10,
                "التفسير الثاني": 10,
                "التجويد": 10,
                "الغياب": 10,
                "السلوك": 10
            }

            maximum = max_scores[exam_name]

            if score < 0 or score > maximum:
                print(f"⚠️ الدرجة يجب أن تكون من 0 إلى {maximum}.")
                return

            index = students.index[
                students["اسم الطالب"] == student_name
            ][0]

            students.loc[index, exam_name] = score

            # حساب المجموع
            total = sum(
                students.loc[index, subject]
                for subject in max_scores
            )

            students.loc[index, "المجموع"] = total
            students.loc[index, "التقدير"] = get_grade(total)

            print("✅ تم حفظ الدرجة بنجاح!")
            print(f"الطالب: {student_name}")
            print(f"{exam_name}: {score} من {maximum}")
            print(f"المجموع: {total}")
            print(f"التقدير: {get_grade(total)}")

    save_btn.on_click(save_grade)
    back_btn.on_click(show_home)

    display(title_grades)
    display(student_box)
    display(exam_box)
    display(score_box)
    display(save_btn)
    display(back_btn)
    display(result)


# ربط زر الدرجات بالواجهة الجديدة
grades_btn._click_handlers.callbacks = []
grades_btn.on_click(show_grades)

# العودة للرئيسية
clear_output(wait=True)
display(title)
display(menu)
```
